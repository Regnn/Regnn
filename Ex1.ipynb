Plot the Corona test results of the training data as a histogram (using 2 bins).

````python
plt.hist(Y_learn, bins=2)
````

What could be a problem in terms of machine learning algorithms to this data? Can you already imagine a technique to counteract this problem?

````python
#We have an imbalanced dataset with more negative than positive results.
#Most machine learning techniques will ignore the minority class and therefore have a poor performance.
````

One approach to address this problem is the Oversampling of the minority class. It is called ```synthetic minority oversampling technique```or short ```SMOTE```. It synthesizes new examples for the minority class from the existing ones.

First import the necessary function.

````python
from imblearn.over_sampling import SMOTE
````

Convert Y_learn to an appropriate data type

````python
Y_learn = np.where(Y_learn > 0.5, 1, 0)
````

Now please apply the ```SMOTE```function to ```X_learn_neu```and ```Y_learn``` to resample the data.

````python
sm = SMOTE()
X_res, Y_res = sm.fit_resample(X_learn_neu, Y_learn)
````

For comparison again plot the resampled Corona test results as a histogram. What has changed?

````python
plt.hist(Y_res, bins=2)
````

````python
#Now there is no imbalance anymore. We have the same amount of negative an positive results.
````
